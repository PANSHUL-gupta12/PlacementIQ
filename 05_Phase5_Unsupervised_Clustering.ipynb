```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score
from scipy.cluster.hierarchy import dendrogram, linkage

sns.set_theme(style="whitegrid", palette="muted")
print("✅ Clustering libraries ready!")
```

```
train = pd.read_csv('dataset/train.csv')
features = ['CGPA', 'Coding_Skills', 'Internships', 'Projects',
            'Communication_Skills', 'Aptitude_Test_Score', 'Soft_Skills_Rating',
            'Certifications', 'Backlogs']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(train[features])
X_sub = X_scaled[:3000] # Subsample for fast evaluation

print("Features scaled and ready.")
```

```
# 1. K-Means Elbow & Silhouette Analysis
k_range = range(2, 7)
inertias, sils = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_sub)
    inertias.append(km.inertia_)
    sils.append(silhouette_score(X_sub, labels))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(k_range, inertias, 'bo-', linewidth=2)
ax1.set_title("Elbow Method (Inertia vs. K)", fontweight='bold')
ax1.set_xlabel("Clusters (K)")
ax1.set_ylabel("Inertia")

ax2.plot(k_range, sils, 'go-', linewidth=2)
ax2.set_title("Silhouette Score vs. K", fontweight='bold')
ax2.set_xlabel("Clusters (K)")
ax2.set_ylabel("Silhouette Score")

plt.tight_layout()
plt.show()
```

```
# 2. Hierarchical Clustering & Dendrogram
Z = linkage(X_scaled[:75], method='ward')

plt.figure(figsize=(14, 5))
dendrogram(Z, truncate_mode='lastp', p=25, leaf_rotation=90.)
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)", fontweight='bold')
plt.xlabel("Sample Index / Cluster Size")
plt.ylabel("Ward Distance")
plt.show()
```

```
# 3. DBSCAN (Density-Based Spatial Clustering with Noise)
db = DBSCAN(eps=2.2, min_samples=15)
db_labels = db.fit_predict(X_sub)

n_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise = list(db_labels).count(-1)

print(f"DBSCAN Results: {n_clusters} clusters identified, {n_noise} noise points detected.")
```
