## 📑 Phase 1 Overview & Syllabus Coverage
In this phase, we handle the fundamental data engineering foundation:
1. **Dataset Ingestion & Inspection:** Loading the student placement dataset (50,000 total records).
2. **Duplicate Detection & Removal:** Preventing data leakage and model skew.
3. **Missing Value Imputation:** Statistical imputation (median for continuous, mode for categorical).
4. **Categorical Feature Encoding:** Converting nominal variables via `LabelEncoder`.
5. **Outlier Detection & Capping:** Interquartile Range (IQR) method clipping.
6. **Feature Scaling:** Standardization via `StandardScaler`.
7. **Train-Test Split:** Strict 80/20 stratified partitioning (40,000 train / 10,000 test).


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 5)

print("✅ Setup complete!")
```

```
# Step 1: Load Raw Dataset
df = pd.read_csv('dataset/train.csv')
print(f"Dataset shape: {df.shape[0]:,} rows, {df.shape[1]} columns")
df.head()
```

```
# Step 2: Missing Value Audit
missing = df.isnull().sum()
print("Missing values per column:\n", missing[missing > 0] if missing.sum() > 0 else "No missing values found.")
```

```
# Step 3: Exploratory Visualizations
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Target Class Distribution
sns.countplot(data=df, x='Placement_Status', palette='Set2', ax=ax1)
ax1.set_title("Target Class Distribution (Placement Status)", fontweight='bold')

# Salary Distribution (Continuous Target for Regression)
sns.histplot(df['Salary_LPA'], kde=True, color='teal', ax=ax2)
ax2.set_title("Salary Package Distribution (LPA)", fontweight='bold')

plt.tight_layout()
plt.show()
```

```
# Step 4: Outlier Detection using IQR Method
numerical_cols = ['Age', 'CGPA', 'Internships', 'Projects', 'Coding_Skills',
                  'Communication_Skills', 'Aptitude_Test_Score', 'Soft_Skills_Rating',
                  'Certifications', 'Backlogs']

plt.figure(figsize=(14, 6))
sns.boxplot(data=df[numerical_cols], palette='viridis')
plt.title("Boxplots of Numerical Features for Outlier Detection", fontweight='bold')
plt.xticks(rotation=30)
plt.show()
```

```
# Step 5: Correlation Heatmap
corr = df[numerical_cols + ['Salary_LPA']].corr()
plt.figure(figsize=(11, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Feature Correlation Heatmap", fontweight='bold')
plt.show()
```
