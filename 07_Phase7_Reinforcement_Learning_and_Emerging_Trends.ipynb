```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (10, 5)
print("✅ RL Environment ready!")
```

```
# Student Preparation MDP & Q-Learning Simulation
class StudentPlacementMDP:
    STATES = {0: "Beginner", 1: "DSA Foundation", 2: "Project Portfolio", 3: "Interview Ready", 4: "Placed (Goal)"}
    ACTIONS = {0: "Solve 50 LeetCode Problems", 1: "Build Full-Stack App", 2: "Soft Skills Training", 3: "Mock Interviews"}
    
    def __init__(self):
        self.state = 0
        self.steps = 0
        
    def reset(self):
        self.state = 0
        self.steps = 0
        return self.state
        
    def step(self, action):
        self.steps += 1
        reward = -1.0 # Effort cost
        p = np.random.rand()
        
        if self.state == 0 and action == 0 and p > 0.2: self.state = 1; reward += 5
        elif self.state == 1 and action == 1 and p > 0.25: self.state = 2; reward += 8
        elif self.state == 2 and action == 2 and p > 0.3: self.state = 3; reward += 10
        elif self.state == 3 and action == 3 and p > 0.2: self.state = 4; reward += 50
            
        done = (self.state == 4) or (self.steps >= 12)
        return self.state, reward, done

env = StudentPlacementMDP()
Q = np.zeros((5, 4))
alpha = 0.1
gamma = 0.95
epsilon = 1.0

rewards = []
for ep in range(800):
    s = env.reset()
    ep_r = 0
    done = False
    while not done:
        a = np.random.randint(4) if np.random.rand() < epsilon else np.argmax(Q[s])
        s_next, r, done = env.step(a)
        best_next_q = 0.0 if done else np.max(Q[s_next])
        Q[s, a] += alpha * (r + gamma * best_next_q - Q[s, a])
        s = s_next
        ep_r += r
    epsilon = max(0.05, epsilon * 0.995)
    rewards.append(ep_r)

plt.figure(figsize=(10, 4))
plt.plot(pd.Series(rewards).rolling(30).mean(), color='navy', linewidth=2)
plt.title("Q-Learning Convergence: Total Episode Reward", fontweight='bold')
plt.xlabel("Training Episode")
plt.ylabel("Rolling Reward")
plt.show()

# Display Optimal Policy
policy = []
for s_id in range(4):
    policy.append({
        "Current Student State": env.STATES[s_id],
        "Optimal Recommended Action": env.ACTIONS[np.argmax(Q[s_id])],
        "Q-Value": f"{np.max(Q[s_id]):.2f}"
    })
print("🎯 Optimal Preparation Policy Learned by RL Agent:")
display(pd.DataFrame(policy))
```

### 📋 Comparative Study: Supervised vs. Unsupervised vs. Reinforcement Learning

| Comparison Criterion | Supervised Learning | Unsupervised Learning | Reinforcement Learning |
| :--- | :--- | :--- | :--- |
| **Input Data** | Labeled data $(X, y)$ | Unlabeled feature vectors $X$ | States, Actions, Rewards |
| **Primary Goal** | Map inputs to target predictions | Discover latent patterns and clusters | Maximize cumulative reward signal |
| **Feedback Mechanism**| Direct error supervision $(y - \hat{y})$ | No external feedback (distance metrics) | Delayed reward scalar from environment |
| **Core Algorithms** | Linear/Logistic Regression, Random Forest, SVM | K-Means, Hierarchical, DBSCAN, PCA, Apriori | Q-Learning, SARSA, DQN, PPO |
| **PlacementIQ Usage**| Classification & Package Regression | Student Segmentation & Skill Itemset Mining | Optimal preparation roadmap navigation |
