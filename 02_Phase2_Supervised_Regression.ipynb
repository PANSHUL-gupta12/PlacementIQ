## 📑 Syllabus Topics Covered:
1. **Regression Concepts & Real-World Applications:** Predicting continuous target variables ($y = f(X)$). In PlacementIQ, predicting student compensation package (in LPA).
2. **Simple Linear Regression:** Ordinary Least Squares (OLS) fitting a single feature line:
   $$y = \beta_0 + \beta_1 X + \epsilon$$
3. **Multiple Linear Regression:** Multi-attribute predictive modeling:
   $$y = \beta_0 + \sum_{j=1}^p \beta_j X_j + \epsilon$$
4. **Polynomial Regression:** Modeling non-linear relationships using higher-degree polynomial features:
   $$y = \beta_0 + \beta_1 X + \beta_2 X^2 + \dots + \beta_d X^d + \epsilon$$
5. **Regression Performance Metrics:**
   - **Mean Absolute Error (MAE):** $\frac{1}{N} \sum_{i=1}^N |y_i - \hat{y}_i|$
   - **Mean Squared Error (MSE):** $\frac{1}{N} \sum_{i=1}^N (y_i - \hat{y}_i)^2$
   - **Root Mean Squared Error (RMSE):** $\sqrt{\text{MSE}}$
   - **$R^2$ Score (Coefficient of Determination):** $1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_theme(style="whitegrid", palette="muted")
print("✅ Libraries imported!")
```

```
# Load the 80/20 train and test sets
train = pd.read_csv('dataset/train.csv')
test = pd.read_csv('dataset/test.csv')

features = ['CGPA', 'Coding_Skills', 'Internships', 'Projects',
            'Communication_Skills', 'Aptitude_Test_Score', 'Soft_Skills_Rating',
            'Certifications', 'Backlogs']
target = 'Salary_LPA'

X_train, y_train = train[features], train[target]
X_test, y_test = test[features], test[target]

print(f"Training on {len(X_train):,} samples (80%), Testing on {len(X_test):,} samples (20%)")
```

```
# 1. Simple Linear Regression (Feature: Coding_Skills)
slr = LinearRegression()
slr.fit(X_train[['Coding_Skills']], y_train)
y_pred_slr = slr.predict(X_test[['Coding_Skills']])

mae_slr = mean_absolute_error(y_test, y_pred_slr)
rmse_slr = np.sqrt(mean_squared_error(y_test, y_pred_slr))
r2_slr = r2_score(y_test, y_pred_slr)

print(f"Simple Linear Regression Results:")
print(f"   • MAE : {mae_slr:.4f} LPA")
print(f"   • RMSE: {rmse_slr:.4f} LPA")
print(f"   • R²  : {r2_slr:.4f}")
```

```
# 2. Multiple Linear Regression (All features)
mlr = LinearRegression()
mlr.fit(X_train, y_train)
y_pred_mlr = mlr.predict(X_test)

mae_mlr = mean_absolute_error(y_test, y_pred_mlr)
rmse_mlr = np.sqrt(mean_squared_error(y_test, y_pred_mlr))
r2_mlr = r2_score(y_test, y_pred_mlr)

print(f"Multiple Linear Regression Results:")
print(f"   • MAE : {mae_mlr:.4f} LPA")
print(f"   • RMSE: {rmse_mlr:.4f} LPA")
print(f"   • R²  : {r2_mlr:.4f}")
```

```
# 3. Polynomial Regression (Degree = 2)
poly_pipe = Pipeline([
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('scaler', StandardScaler()),
    ('ridge', Ridge(alpha=1.0))
])

poly_pipe.fit(X_train, y_train)
y_pred_poly = poly_pipe.predict(X_test)

mae_poly = mean_absolute_error(y_test, y_pred_poly)
rmse_poly = np.sqrt(mean_squared_error(y_test, y_pred_poly))
r2_poly = r2_score(y_test, y_pred_poly)

print(f"Polynomial Regression (Degree 2) Results:")
print(f"   • MAE : {mae_poly:.4f} LPA")
print(f"   • RMSE: {rmse_poly:.4f} LPA")
print(f"   • R²  : {r2_poly:.4f}")
```

```
# 4. Regression Model Comparison Table & Visualizations
comparison = pd.DataFrame([
    {"Model": "Simple Linear Regression", "MAE": mae_slr, "RMSE": rmse_slr, "R2_Score": r2_slr},
    {"Model": "Multiple Linear Regression", "MAE": mae_mlr, "RMSE": rmse_mlr, "R2_Score": r2_mlr},
    {"Model": "Polynomial Regression (Deg 2)", "MAE": mae_poly, "RMSE": rmse_poly, "R2_Score": r2_poly},
])
display(comparison)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Actual vs Predicted
ax1.scatter(y_test[:250], y_pred_mlr[:250], alpha=0.6, color='teal')
ax1.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', label='Perfect Fit')
ax1.set_title("Multiple Linear Regression: Actual vs. Predicted (LPA)", fontweight='bold')
ax1.set_xlabel("Actual Salary")
ax1.set_ylabel("Predicted Salary")
ax1.legend()

# Residual Plot
residuals = y_test - y_pred_mlr
sns.histplot(residuals, kde=True, color='purple', ax=ax2)
ax2.set_title("Residual Error Distribution (Actual - Predicted)", fontweight='bold')
ax2.axvline(0, color='red', linestyle='--')

plt.tight_layout()
plt.show()
```
